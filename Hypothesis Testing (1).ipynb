## The Poisson distribution

If $X$ follows a Poisson distribution with parameter $\lambda$, then the probability mass function (PMF) of $X$ is given by:

$$P(X = x,\lambda) = \frac{e^{-\lambda} \lambda^x}{x!}$$

Where:
\begin{align*}
&\lambda \text{ is the parameter of the Poisson distribution,} \\
&x \text{ is a non-negative integer,} \\
&x! \text{ denotes the factorial of } x.
\end{align*}

We now consider $X_1, X_2, \ldots, X_n$ as $n$ independent and identically distributed (i.i.d.) random variables following the Poisson PMF $X_i\sim P(\lambda)$.
